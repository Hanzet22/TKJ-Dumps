```python
# ================================================
#  🕵️ FurTrace OSINT Lite v1.0
#  Produk TKJ - Karya: Resti
#  Fungsi: OSINT sederhana untuk website
# ================================================

import socket
import whois
import dns.resolver
import requests
from datetime import datetime

print("""
   ╔═══════════════════════════════════════╗
   ║   🕵️  FurTrace OSINT Lite v1.0       ║
   ║   Open Source Intelligence Sederhana  ║
   ╚═══════════════════════════════════════╝
""")

target = input("🎯 Masukkan domain (contoh: google.com): ").strip()

# 1. IP Address
try:
    ip = socket.gethostbyname(target)
    print(f"\n📡 IP Address: {ip}")
except:
    print(f"\n❌ Gagal resolve domain")

# 2. WHOIS
try:
    w = whois.whois(target)
    print(f"\n📝 WHOIS Info:")
    print(f"   Registrar: {w.registrar}")
    print(f"   Created: {w.creation_date}")
    print(f"   Expires: {w.expiration_date}")
except:
    print(f"\n⚠️ WHOIS tidak tersedia")

# 3. DNS Records
try:
    print(f"\n📡 DNS Records:")
    for record in ['A', 'MX', 'NS', 'TXT']:
        try:
            answers = dns.resolver.resolve(target, record)
            for r in answers:
                print(f"   {record}: {r}")
        except:
            pass
except:
    pass

# 4. Cek HTTP Header
try:
    r = requests.get(f"http://{target}", timeout=5, allow_redirects=True)
    print(f"\n🌐 HTTP Info:")
    print(f"   Status: {r.status_code}")
    print(f"   Server: {r.headers.get('Server', 'N/A')}")
    print(f"   Redirect: {r.url}")
except:
    pass

print("\n✨ OSINT Scan Selesai!")
```
