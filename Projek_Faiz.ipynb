```python

# ================================================
#  📊 WebRec Analyzer v1.0
#  Produk TKJ - Karya: Faiz
#  Fungsi: Analisis website langsung
# ================================================

import requests
import json
from collections import Counter
from urllib.parse import urlparse

print("""
   ╔═══════════════════════════════════════╗
   ║   📊  WebRec Analyzer v1.0 (SIMPEL)  ║
   ║   Analisis Web Tanpa Upload          ║
   ╚═══════════════════════════════════════╝
""")

target = input("🌐 Masukkan URL (contoh: https://google.com): ").strip()

print("\n⏳ Menganalisis...")

try:
    # Ambil halaman
    r = requests.get(target, timeout=10, allow_redirects=True)

    print(f"\n📊 Statistik:")
    print(f"   Status Code: {r.status_code}")
    print(f"   Redirect: {r.url}")
    print(f"   Content Length: {len(r.text)} karakter")

    # Header
    print(f"\n📡 Header:")
    for key, value in r.headers.items():
        print(f"   {key}: {value}")

    # Link counter
    import re
    links = re.findall(r'href=[\'"]?([^\'" >]+)', r.text)
    internal = []
    external = []

    for link in links:
        if link.startswith('http'):
            if urlparse(link).netloc == urlparse(target).netloc:
                internal.append(link)
            else:
                external.append(link)
        elif link.startswith('/') or not link.startswith('http'):
            internal.append(link)

    print(f"\n🔗 Link Count:")
    print(f"   Total: {len(links)}")
    print(f"   Internal: {len(internal)}")
    print(f"   External: {len(external)}")

    # Security headers
    print(f"\n🛡️ Security Headers:")
    headers = r.headers
    hsts = headers.get('Strict-Transport-Security', '❌')
    xframe = headers.get('X-Frame-Options', '❌')
    csp = headers.get('Content-Security-Policy', '❌')
    print(f"   HSTS: {hsts}")
    print(f"   X-Frame-Options: {xframe}")
    print(f"   CSP: {csp}")

    print("\n✨ Analisis selesai!")

except Exception as e:
    print(f"❌ Error: {e}")
```
