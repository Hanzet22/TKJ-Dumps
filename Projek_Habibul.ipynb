```python


# ========================================
#  🔒 PortGuard Hybrid v1.0
#  Produk TKJ - Karya: Habibul
#  Fungsi: Scan Port + Deteksi Layanan
#  Mode: Basic & Advanced
# ========================================

import socket
import time
import platform

print("""
   ╔═══════════════════════════════════════╗
   ║   🔒  PortGuard Hybrid v1.0          ║
   ║   Scan Port & Deteksi Layanan        ║
   ║   Mode: Basic | Advanced             ║
   ╚═══════════════════════════════════════╝
""")

# ===== DATABASE PORT =====
port_db = {
    20: "FTP-DATA", 21: "FTP", 22: "SSH", 23: "Telnet",
    25: "SMTP", 53: "DNS", 67: "DHCP", 68: "DHCP",
    69: "TFTP", 80: "HTTP", 110: "POP3", 111: "RPC",
    123: "NTP", 135: "RPC", 137: "NetBIOS", 138: "NetBIOS",
    139: "NetBIOS", 143: "IMAP", 161: "SNMP", 162: "SNMP",
    179: "BGP", 389: "LDAP", 443: "HTTPS", 445: "SMB",
    465: "SMTPS", 514: "Syslog", 515: "LPD", 587: "SMTP",
    636: "LDAPS", 873: "RSYNC", 993: "IMAPS", 995: "POP3S",
    1080: "SOCKS", 1433: "MSSQL", 1521: "Oracle", 1723: "PPTP",
    1883: "MQTT", 2049: "NFS", 2086: "WHM", 2087: "WHM",
    2181: "ZooKeeper", 2222: "SSH-Alt", 2375: "Docker", 2376: "Docker",
    2424: "OrientDB", 3000: "Grafana", 3306: "MySQL", 3389: "RDP",
    3632: "DistCC", 4243: "Docker", 4443: "HTTPS-Alt", 5000: "Flask",
    5222: "XMPP", 5223: "XMPP", 5432: "PostgreSQL", 5555: "ADB",
    5666: "NRPE", 5672: "RabbitMQ", 5984: "CouchDB", 6000: "X11",
    6379: "Redis", 6443: "Kubernetes", 6881: "BitTorrent", 7001: "WebLogic",
    7200: "OpenNMS", 7300: "OpenNMS", 7474: "Neo4j", 7547: "TR-069",
    7777: "Oracle XE", 8000: "HTTP-Alt", 8080: "HTTP-Alt", 8081: "HTTP-Alt",
    8086: "InfluxDB", 8089: "Splunk", 8090: "HTTP-Alt", 8091: "Couchbase",
    8140: "Puppet", 8161: "ActiveMQ", 8181: "HTTP-Alt", 8200: "Minio",
    8443: "HTTPS-Alt", 8500: "Consul", 8600: "DNS-Alt", 8883: "MQTT-SSL",
    8888: "HTTP-Alt", 9000: "Jetty", 9042: "Cassandra", 9092: "Kafka",
    9094: "Kafka-SSL", 9100: "NodeExport", 9200: "Elasticsearch", 9300: "Elasticsearch",
    9418: "Git", 9999: "HTTP-Alt", 10000: "WebMin", 11111: "HTTP-Alt",
    15672: "RabbitMQ", 16000: "Hadoop", 20000: "Druid", 27017: "MongoDB",
    50000: "SAP", 55000: "SAP", 60000: "SAP"
}

# ===== FUNGSI SCAN =====
def scan_port(target, port):
    try:
        sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
        sock.settimeout(1.5)
        hasil = sock.connect_ex((target, port))
        sock.close()
        return hasil == 0
    except:
        return False

# ===== INPUT TARGET =====
target = input("🎯 Masukkan IP/Domain target: ")
print(f"\n📡 Target: {target}")
print(f"💻 OS: {platform.system()} {platform.release()}")

# ===== PILIH MODE =====
print("\n📋 Pilih Mode Scan:")
print("   [1] BASIC - 20 port umum (cepat)")
print("   [2] ADVANCED - 80 port populer + layanan")
print("   [3] CUSTOM - Input port sendiri")

mode = input("\nPilih (1/2/3): ")

# ===== TENTUKAN PORT =====
if mode == "1":
    port_list = [21, 22, 23, 25, 53, 80, 110, 139, 143, 443,
                 445, 993, 995, 1433, 1521, 3306, 3389, 5432,
                 6379, 27017]
    mode_name = "BASIC (20 port)"
elif mode == "2":
    port_list = list(port_db.keys())[:80]  # 80 port pertama
    mode_name = "ADVANCED (80 port)"
elif mode == "3":
    custom = input("Masukkan port (pisah koma, contoh: 80,443,3306): ")
    port_list = [int(p.strip()) for p in custom.split(',')]
    mode_name = f"CUSTOM ({len(port_list)} port)"
else:
    print("❌ Pilihan salah! Pakai mode BASIC")
    port_list = [21, 22, 23, 25, 53, 80, 110, 139, 143, 443,
                 445, 993, 995, 1433, 1521, 3306, 3389, 5432,
                 6379, 27017]
    mode_name = "BASIC (20 port)"

print(f"\n⏳ Scanning {mode_name}...")
print("="*45)

# ===== PROSES SCAN =====
open_ports = []
total = len(port_list)
found = 0

for idx, port in enumerate(port_list):
    # Progress bar
    progress = int((idx + 1) / total * 100)
    bar = "█" * (progress // 2) + "░" * (50 - progress // 2)
    print(f"\r   Progress: [{bar}] {progress}%", end="", flush=True)

    if scan_port(target, port):
        found += 1
        service = port_db.get(port, "Unknown")
        open_ports.append((port, service))
        print(f"\r   ✅ PORT {port} TERBUKA → {service}")

print("\n" + "="*45)

# ===== HASIL SCAN =====
print(f"\n📊 HASIL SCAN:")
print(f"   Total port discan: {total}")
print(f"   Port terbuka: {found}")

if open_ports:
    print("\n🔓 PORT TERBUKA:")
    for port, service in open_ports:
        print(f"   ✅ {port} → {service}")

    # ===== ANALISA KEAMANAN =====
    print("\n🛡️ ANALISA KEAMANAN:")

    critical_ports = [21, 23, 80, 443, 3389, 3306, 445, 22]
    critical_found = [p for p, s in open_ports if p in critical_ports]

    if not critical_found:
        print("   ✅ AMAN! Tidak ada port kritis terbuka")
    else:
        print("   ⚠️ PERINGATAN! Ada port kritis terbuka:")
        for port in critical_found:
            service = port_db.get(port, "Unknown")
            print(f"      - Port {port} ({service})")

        print("\n   💡 REKOMENDASI:")
        if 21 in critical_found:
            print("      - FTP (21): Matikan kalo gak dipake")
        if 23 in critical_found:
            print("      - Telnet (23): Ganti dengan SSH")
        if 22 in critical_found:
            print("      - SSH (22): Gunakan key authentication")
        if 3389 in critical_found:
            print("      - RDP (3389): Batasi akses IP")
        if 3306 in critical_found:
            print("      - MySQL (3306): Restrict IP atau bind local")
        if 445 in critical_found:
            print("      - SMB (445): Matikan kalo gak perlu")
        if 80 in critical_found or 443 in critical_found:
            print("      - HTTP/HTTPS (80/443): Update web server secara rutin")

    # ===== SUGESTI =====
    print("\n💡 SUGESTI:")
    if found <= 3:
        print("   ✅ Server cukup aman")
    elif found <= 6:
        print("   ⚠️ Coba kurangi port yang tidak perlu")
    else:
        print("   ❌ Terlalu banyak port terbuka! Segera lockdown!")

else:
    print("\n✅ Tidak ada port terbuka!")
    print("💡 Server sangat aman!")

# ===== INFO TAMBAHAN =====
print(f"\n📋 Waktu scan: {time.strftime('%d/%m/%Y %H:%M:%S')}")
print(f"🎯 Target: {target}")
print(f"📋 Mode: {mode_name}")

# Cek koneksi target
print("\n📡 STATUS TARGET:")
try:
    ip = socket.gethostbyname(target)
    print(f"   ✅ IP Resolved: {ip}")
except:
    print("   ❌ Domain tidak bisa di-resolve")

print("\n" + "="*40)
print("📋 Laporan scan selesai!")
print("✨ Produk TKJ by Habibul")
```
