```python
# ================================================
#  🔗 FindLink Pro v1.0
#  Produk TKJ - Karya: Qeyla
#  Fungsi: Mencari semua link di website
# ================================================

import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin, urlparse

print("""
   ╔═══════════════════════════════════════╗
   ║   🔗  FindLink Pro v1.0              ║
   ║   Web Link Finder & Scraper          ║
   ╚═══════════════════════════════════════╝
""")

target = input("🎯 Masukkan URL (contoh: https://google.com): ").strip()

try:
    r = requests.get(target, timeout=10)
    soup = BeautifulSoup(r.text, 'html.parser')

    # Ambil semua link
    links = soup.find_all('a', href=True)

    internal = []
    external = []

    for link in links:
        url = urljoin(target, link['href'])
        if urlparse(url).netloc == urlparse(target).netloc:
            internal.append(url)
        else:
            external.append(url)

    print(f"\n📊 Hasil Scan:")
    print(f"   Total Link: {len(links)}")
    print(f"   Internal: {len(internal)}")
    print(f"   External: {len(external)}")

    print(f"\n🔗 Internal Links (5 sample):")
    for link in internal[:5]:
        print(f"   ➜ {link}")

    print(f"\n🌐 External Links (5 sample):")
    for link in external[:5]:
        print(f"   ➜ {link}")

except Exception as e:
    print(f"❌ Error: {e}")
    print("💡 Install: pip install beautifulsoup4 requests")
```
