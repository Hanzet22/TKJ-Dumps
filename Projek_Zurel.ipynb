```python
# ================================================
#  📹 SeariTube v1.0
#  Produk TKJ - Karya: Zurel
#  Fungsi: Pencarian YouTube sederhana
# ================================================

print("""
   ╔═══════════════════════════════════════╗
   ║   📹  SeariTube v1.0                 ║
   ║   YouTube Searcher Sederhana          ║
   ╚═══════════════════════════════════════╝
""")

query = input("🔍 Masukkan kata kunci: ").strip()

# Pake yt-dlp atau scrap manual
# Karena butuh API key, kita pake alternatif

print(f"\n📊 Hasil pencarian untuk: '{query}'")
print("   (Menggunakan YouTube HTML scraper)")

import requests
from urllib.parse import quote

try:
    # Scrape hasil pencarian YouTube (tanpa API)
    search_url = f"https://www.youtube.com/results?search_query={quote(query)}"
    headers = {'User-Agent': 'Mozilla/5.0'}
    r = requests.get(search_url, headers=headers, timeout=10)

    # Parse video ID
    import re
    video_ids = re.findall(r'"videoId":"([^"]+)"', r.text)
    titles = re.findall(r'"title":{"runs":\[{"text":"([^"]+)"', r.text)

    print(f"\n📹 Hasil (5 video pertama):")
    for i in range(min(5, len(video_ids))):
        video_id = video_ids[i] if i < len(video_ids) else "N/A"
        title = titles[i] if i < len(titles) else "Unknown"
        url = f"https://youtube.com/watch?v={video_id}"
        print(f"   {i+1}. {title[:50]}")
        print(f"      🔗 {url}")

except Exception as e:
    print(f"❌ Error: {e}")
    print("💡 Alternatif: pake yt-dlp atau API key")
```
