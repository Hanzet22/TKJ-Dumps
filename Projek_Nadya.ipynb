```python
# ================================================
#  ⚡ SpeedPulse v1.0
#  Produk TKJ - Karya: Nadya
#  Fungsi: Tes Kecepatan Internet (Download/Upload/Ping)
# ================================================

# Install speedtest-cli (khusus Colab)
!pip install -q speedtest-cli

import speedtest
import time
from datetime import datetime

print("""
   ╔═══════════════════════════════════════╗
   ║   ⚡  SpeedPulse v1.0                ║
   ║   Tes Kecepatan Internet Simpel      ║
   ╚═══════════════════════════════════════╝
""")

print("⏳ Mempersiapkan test...")
print("📡 Mencari server terbaik...")

st = speedtest.Speedtest()

# Cari server terbaik
st.get_best_server()
server_info = st.results.server

print(f"\n📍 Server: {server_info.get('name', 'Unknown')} ({server_info.get('country', 'Unknown')})")
print(f"   Host: {server_info.get('host', 'N/A')}")
print(f"   Jarak: {server_info.get('latency', 0):.0f} ms")

print("\n⏳ Mengukur kecepatan...")
print("   ⬇️ Download: 0%", end="", flush=True)

# Test download
download = st.download() / 1_000_000  # Mbps
print(f"\r   ⬇️ Download: 100% — {download:.2f} Mbps")

print("   ⬆️ Upload: 0%", end="", flush=True)

# Test upload
upload = st.upload() / 1_000_000  # Mbps
print(f"\r   ⬆️ Upload: 100% — {upload:.2f} Mbps")

# Ping
ping = st.results.ping

print("\n" + "="*50)
print("📊 HASIL TEST:")
print(f"   ⬇️ Download : {download:.2f} Mbps")
print(f"   ⬆️ Upload   : {upload:.2f} Mbps")
print(f"   🏓 Ping     : {ping:.0f} ms")
print("="*50)

# ===== REKOMENDASI =====
print("\n💡 REKOMENDASI:")

if download >= 100:
    print("   🚀 INTERNET SANGAT CEPAT!")
    print("   ✅ Cocok buat:")
    print("      - Gaming online (game berat)")
    print("      - Streaming 4K")
    print("      - Download file besar")
    print("      - Multiple devices")
elif download >= 50:
    print("   ⚡ INTERNET CEPAT!")
    print("   ✅ Cocok buat:")
    print("      - Streaming HD/4K")
    print("      - Gaming online")
    print("      - Browsing cepat")
    print("      - Video call HD")
elif download >= 25:
    print("   👍 INTERNET SEDANG!")
    print("   ✅ Cocok buat:")
    print("      - Streaming HD")
    print("      - Browsing")
    print("      - Video call")
    print("      - Download file")
elif download >= 10:
    print("   ⚠️ INTERNET LAMBAT!")
    print("   ✅ Cocok buat:")
    print("      - Browsing basic")
    print("      - YouTube 480p")
    print("      - Chatting")
    print("   💡 Saran: Upgrade paket internet")
else:
    print("   ❌ INTERNET SANGAT LAMBAT!")
    print("   💡 Saran:")
    print("      - Cek sinyal WiFi")
    print("      - Restart router")
    print("      - Hubungi provider")

# ===== KUALITAS =====
print("\n📊 KUALITAS INTERNET:")
if ping < 20:
    print("   ✅ Ping sangat rendah (bagus buat gaming)")
elif ping < 50:
    print("   ✅ Ping rendah (bagus buat streaming)")
elif ping < 100:
    print("   ⚠️ Ping sedang (masih ok buat browsing)")
else:
    print("   ❌ Ping tinggi (lemot, cocok buat kirim chat doang)")

print(f"\n📋 Waktu test: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
print(f"📍 Server test: {server_info.get('name', 'N/A')}, {server_info.get('country', 'N/A')}")
print("\n✨ Tes selesai!")
```
