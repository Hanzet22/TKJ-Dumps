```python

# ================================================
#  🔒 CertiSafe v1.0 — FIXED
#  Produk TKJ - Karya: Iwan
#  Fungsi: Cek Sertifikat SSL/TLS
# ================================================

import ssl
import socket
from datetime import datetime
import dateutil.parser

print("""
   ╔═══════════════════════════════════════╗
   ║   🔒  CertiSafe v1.0                 ║
   ║   SSL/TLS Certificate Checker        ║
   ╚═══════════════════════════════════════╝
""")

target = input("🎯 Masukkan domain: ").strip()

try:
    context = ssl.create_default_context()
    with socket.create_connection((target, 443), timeout=5) as sock:
        with context.wrap_socket(sock, server_hostname=target) as ssock:
            cert = ssock.getpeercert()

            print(f"\n🔒 Certificate Info:")
            print(f"   Subject: {cert.get('subject', [])[0]}")
            print(f"   Issuer: {cert.get('issuer', [])[0]}")
            print(f"   Valid From: {cert.get('notBefore')}")
            print(f"   Valid Until: {cert.get('notAfter')}")

            # ===== FIX: PARSING TIMEZONE =====
            try:
                # Parse expire date
                expire = dateutil.parser.parse(cert['notAfter'])

                # Buat waktu sekarang dengan timezone UTC
                now = datetime.now(dateutil.tz.tzutc())

                # Hitung selisih
                days_left = (expire - now).days

                print(f"\n⏳ Masa Berlaku:")
                if days_left > 30:
                    print(f"   ✅ {days_left} hari tersisa (Aman)")
                elif days_left > 7:
                    print(f"   ⚠️ {days_left} hari tersisa (Segera perpanjang)")
                else:
                    print(f"   ❌ {days_left} hari tersisa (KRITIS!)")
            except:
                print(f"\n⏳ Masa Berlaku: (Gagal hitung)")

            print(f"\n🔐 TLS Version: {ssock.version()}")
            print(f"🔑 Cipher: {ssock.cipher()[0]}")

except Exception as e:
    print(f"❌ Error: {e}")
    print("💡 Pastikan domain punya SSL/HTTPS")
```
