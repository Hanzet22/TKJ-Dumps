```python
# ========================================
#  🌐 IPTracker v1.0
#  Produk TKJ - Karya: Richardo
#  Fungsi: Lacak IP Lokal & Publik
# ========================================

import socket
import urllib.request
import platform
import json
import time

print("""
   ╔═══════════════════════════════════╗
   ║   🌐  IPTracker v1.0             ║
   ║   Lacak Alamat IP & Lokasi       ║
   ╚═══════════════════════════════════╝
""")

print("⏳ Mengumpulkan data jaringan...")
time.sleep(1)

# ===== 1. INFO SISTEM =====
print("\n📊 INFORMASI SISTEM:")
print(f"   💻 OS      : {platform.system()} {platform.release()}")
print(f"   🖥️  Device  : {platform.node()}")
print(f"   🏗️  Arsitektur: {platform.machine()}")

# ===== 2. IP LOKAL =====
try:
    hostname = socket.gethostname()
    ip_lokal = socket.gethostbyname(hostname)
    print(f"\n🏠 IP LOKAL:")
    print(f"   ➜ {ip_lokal}")
    print("   (Digunakan di jaringan lokal/WiFi)")
except:
    print("\n🏠 IP LOKAL:")
    print("   ❌ Gagal mendapatkan IP lokal")

# ===== 3. IP PUBLIK =====
print("\n⏳ Mengambil IP Publik...")
try:
    # Ambil IP publik dari api
    ip_publik = urllib.request.urlopen('https://api.ipify.org').read().decode()
    print(f"\n🌍 IP PUBLIK:")
    print(f"   ➜ {ip_publik}")
    print("   (Digunakan di internet)")

    # ===== 4. INFO LOKASI (BONUS) =====
    print("\n⏳ Mencari lokasi IP...")
    try:
        # Ambil info lokasi dari ip-api.com (gratis, no API key)
        url = f'http://ip-api.com/json/{ip_publik}'
        response = urllib.request.urlopen(url)
        data = json.loads(response.read().decode())

        if data['status'] == 'success':
            print("\n📍 LOKASI IP:")
            print(f"   🌏 Negara  : {data['country']}")
            print(f"   🏙️  Kota    : {data['city']}")
            print(f"   📍 Region  : {data['regionName']}")
            print(f"   📮 Kode Pos: {data['zip']}")
            print(f"   📡 ISP     : {data['isp']}")
            print(f"   🗺️  Koordinat: {data['lat']}, {data['lon']}")
        else:
            print("   ❌ Gagal dapat lokasi")
    except:
        print("   ❌ Gagal dapat lokasi (cek internet)")

except:
    print("\n🌍 IP PUBLIK:")
    print("   ❌ Gagal mendapatkan IP Publik")
    print("   💡 Cek koneksi internet!")

# ===== 5. INFO JARINGAN TAMBAHAN =====
print("\n📡 INFO JARINGAN TAMBAHAN:")
try:
    # Cek DNS
    socket.gethostbyname('google.com')
    print("   ✅ DNS: Berfungsi")
except:
    print("   ❌ DNS: Gagal")

# Cek koneksi internet (pake socket)
try:
    sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
    sock.settimeout(2)
    sock.connect(('8.8.8.8', 53))
    sock.close()
    print("   ✅ Internet: Terhubung")
except:
    print("   ❌ Internet: Tidak terhubung")

print("\n" + "="*40)
print("📋 Laporan IP selesai!")
print("✨ Produk TKJ by Richardo")
```
