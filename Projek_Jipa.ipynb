```python
# ================================================
#  🕵️ Username Hunter v1.0
#  Produk TKJ - Karya: Jipa
#  Fungsi: Mencari username di berbagai platform
# ================================================

import requests
import json
import time
from concurrent.futures import ThreadPoolExecutor, as_completed

print("""
   ╔═══════════════════════════════════════╗
   ║   🕵️  Username Hunter v1.0           ║
   ║   Cari Username di 10+ Platform      ║
   ╚═══════════════════════════════════════╝
""")

# Platform yang dicek
PLATFORMS = {
    'GitHub': 'https://github.com/{}',
    'Twitter/X': 'https://twitter.com/{}',
    'Instagram': 'https://instagram.com/{}',
    'YouTube': 'https://youtube.com/@{}',
    'TikTok': 'https://tiktok.com/@{}',
    'Reddit': 'https://reddit.com/user/{}',
    'Medium': 'https://medium.com/@{}',
    'Dev.to': 'https://dev.to/{}',
    'Pinterest': 'https://pinterest.com/{}',
    'Spotify': 'https://open.spotify.com/user/{}',
    'Twitch': 'https://twitch.tv/{}',
    'Steam': 'https://steamcommunity.com/id/{}',
    'Patreon': 'https://patreon.com/{}',
    'SoundCloud': 'https://soundcloud.com/{}',
    'Vimeo': 'https://vimeo.com/{}',
}

def check_username(platform, username):
    """Cek username di satu platform"""
    url = PLATFORMS[platform].format(username)
    try:
        response = requests.get(url, timeout=5)
        if response.status_code == 200:
            return platform, True, url
        else:
            return platform, False, url
    except:
        return platform, False, url

print("🎯 Masukkan username: ")
username = input("➜ ").strip()

if not username:
    print("❌ Username tidak boleh kosong!")
    exit()

print(f"\n⏳ Mencari username '{username}' di {len(PLATFORMS)} platform...\n")

found = []
not_found = []

with ThreadPoolExecutor(max_workers=5) as executor:
    futures = {executor.submit(check_username, platform, username): platform for platform in PLATFORMS}

    for future in as_completed(futures):
        platform, is_found, url = future.result()
        if is_found:
            print(f"✅ {platform}: {url}")
            found.append((platform, url))
        else:
            print(f"❌ {platform}: Tidak ditemukan")
            not_found.append(platform)

print("\n" + "="*50)
print("📊 HASIL PENCARIAN:")
print(f"   Ditemukan: {len(found)} platform")
print(f"   Tidak ditemukan: {len(not_found)} platform")

if found:
    print("\n🔗 LINK AKTIF:")
    for platform, url in found:
        print(f"   ➜ {platform}: {url}")

print("\n✨ Pencarian selesai!")
```
