```python
# ================================================
#  👁️ WebSight Scanner v1.0
#  Produk TKJ - Karya: Nadin
#  Fungsi: Scan Web sederhana
# ================================================

import socket
import requests

print("""
   ╔═══════════════════════════════════════╗
   ║   👁️  WebSight Scanner v1.0          ║
   ║   Scan Keamanan Web Sederhana         ║
   ╚═══════════════════════════════════════╝
""")

target = input("🎯 Masukkan domain: ").strip()

# 1. Cek Port 80 & 443
ports = [80, 443]
open_ports = []

print(f"\n🔍 Scanning {target}...")
for port in ports:
    try:
        sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
        sock.settimeout(2)
        result = sock.connect_ex((target, port))
        sock.close()
        if result == 0:
            open_ports.append(port)
            print(f"   ✅ Port {port} terbuka")
        else:
            print(f"   ❌ Port {port} tertutup")
    except:
        print(f"   ❌ Port {port} error")

# 2. Cek HTTPS
print(f"\n🔒 SSL/TLS Check:")
try:
    r = requests.get(f"https://{target}", timeout=5)
    print(f"   ✅ HTTPS aktif (Status: {r.status_code})")
    print(f"   🔐 Redirect: {r.url}")
except:
    print(f"   ❌ HTTPS tidak aktif")

# 3. Cek Security Headers
print(f"\n🛡️ Security Headers:")
try:
    r = requests.get(f"https://{target}", timeout=5, allow_redirects=True)
    headers = r.headers

    hsts = headers.get('Strict-Transport-Security', '❌ Tidak ada')
    print(f"   HSTS: {hsts}")

    xframe = headers.get('X-Frame-Options', '❌ Tidak ada')
    print(f"   X-Frame-Options: {xframe}")

    server = headers.get('Server', '❌ Tidak ada')
    print(f"   Server: {server}")
except:
    print(f"   ❌ Gagal mengambil header")

print(f"\n📊 Port terbuka: {len(open_ports)}")
```
