```python

# ========================================
#  📡 WiFiScope v1.0 (Versi Python)
#  Produk TKJ - Karya: Marwa
#  Fungsi: Cek Koneksi Jaringan
# ========================================

import socket
import platform
import time

print("""
   ╔═══════════════════════════════════╗
   ║   📡  WiFiScope v1.0             ║
   ║   Analisa Koneksi Jaringan       ║
   ╚═══════════════════════════════════╝
""")

print("⏳ Memeriksa koneksi...")
time.sleep(1)

# Cek koneksi ke beberapa server (pake socket, gak perlu ping)
server_list = [
    ("google.com", 80),
    ("8.8.8.8", 53),
    ("1.1.1.1", 53)
]

ketemu = False

for server, port in server_list:
    try:
        sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
        sock.settimeout(2)
        hasil = sock.connect_ex((server, port))
        sock.close()

        if hasil == 0:
            print(f"\n✅ Terhubung ke: {server}")
            print("📶 Status: KONEKSI AKTIF")
            print("💡 Internet: Bisa akses")
            ketemu = True
            break
    except:
        continue

if not ketemu:
    print("\n❌ Tidak ada koneksi internet")
    print("💡 Solusi:")
    print("   1. Cek WiFi menyala")
    print("   2. Cek kabel LAN")
    print("   3. Cek router")

# Info tambahan
print(f"\n🖥️  Perangkat: {platform.node()}")
print(f"💻 OS: {platform.system()} {platform.release()}")

# Cek IP lokal
try:
    hostname = socket.gethostname()
    ip_lokal = socket.gethostbyname(hostname)
    print(f"🏠 IP Lokal: {ip_lokal}")
except:
    pass

print("\n📋 Laporan selesai!")
```
